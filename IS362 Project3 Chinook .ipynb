This notebook uses SQLite and pandas to combine customer purchase information from five tables in the Chinook database.

## 1. Database and required libraries

I chose **SQLite** because it is simple to use and Python already includes the `sqlite3` middleware library. The database connection uses the local SQLite file path. If the database file is not present, the code downloads it from the official Chinook GitHub release.

```python
from pathlib import Path
from urllib.request import urlretrieve
import sqlite3
import pandas as pd

database_path = Path('Chinook_Sqlite.sqlite')
database_url = ('https://github.com/lerocha/chinook-database/'
                'releases/download/v1.4.5/Chinook_Sqlite.sqlite')

if not database_path.exists():
    urlretrieve(database_url, database_path)

print(f'Database ready: {database_path}')
```

## 2. Connect to the database

The connection string for SQLite is the database file name. The `sqlite3.connect()` function opens the database so that pandas can run the SQL query.

```python
connection = sqlite3.connect(database_path)
print('Connected to the Chinook database.')
```

## 3. SQL query joining five tables

The query joins these five tables: `Customer`, `Invoice`, `InvoiceLine`, `Track`, and `Album`. It returns each customer's last name, first name, purchased track name, and album title. The results are sorted by customer last name and then first name.

```python
query = '''
SELECT
    Customer.LastName AS LastName,
    Customer.FirstName AS FirstName,
    Track.Name AS TrackName,
    Album.Title AS AlbumTitle
FROM Customer
JOIN Invoice
    ON Customer.CustomerId = Invoice.CustomerId
JOIN InvoiceLine
    ON Invoice.InvoiceId = InvoiceLine.InvoiceId
JOIN Track
    ON InvoiceLine.TrackId = Track.TrackId
JOIN Album
    ON Track.AlbumId = Album.AlbumId
ORDER BY Customer.LastName, Customer.FirstName;
'''

purchases = pd.read_sql_query(query, connection)
purchases.head()
```

## 4. Review the completed DataFrame

The DataFrame contains one row for every purchased track. The first five rows begin with customer Roberto Almeida, which confirms that the requested sorting was applied.

```python
print(f'Total rows: {len(purchases):,}')
print(f'Total columns: {purchases.shape[1]}')
purchases.head(10)
```

```python
connection.close()
print('Database connection closed.')
```

## Conclusion

The SQL query successfully combined information from five related tables and loaded the results into a pandas DataFrame. The final output shows customer names, purchased tracks, and album titles in the required order.
